```python
import cv2
from IPython.display import clear_output
import matplotlib.pyplot as plt
from numpy.random import randn
from time import sleep
from matplotlib import pyplot as pyplot
import numpy as np


from google.colab.patches import cv2_imshow
finalcount = 0
video = cv2.VideoCapture('/content/pos/posVideo1.868.avi')

frame_width = int(video.get(3))
frame_height = int(video.get(4))
frame_size = (frame_width,frame_height)
fps = 20
output = cv2.VideoWriter('output_video.avi', cv2.VideoWriter_fourcc(*'MJPG'), fps, frame_size)

while (video.isOpened()):
  ret, frame = video.read()
  if ret == True: 
    grayvideo = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edge_frame = cv2.Canny(grayvideo, 250, 400)

    BGRvideo = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    hsv = cv2.cvtColor(BGRvideo, cv2.COLOR_RGB2HSV)  
    lower_value = np.array([0,125,125])
    upper_value = np.array([40,255,255])

    mask = cv2.inRange(hsv, lower_value, upper_value)
    res = cv2.bitwise_and(frame,frame, mask=mask)

    output.write(cv2.cvtColor(edge_frame, cv2.COLOR_GRAY2BGR))
    output.write(cv2.cvtColor(res, cv2.COLOR_HSV2BGR))

    count = cv2.countNonZero(mask)

    if int(count) > 10000:
      finalcount = 1
      
  else:
    break
if finalcount == 1:
  print('Fire Detected')
else:
  print('No Fire Detected')
```
